# Dark Fleets: Dataset Exploration

This notebook loads the primary Global Fishing Watch dataset (AIS Disabling Events) and the secondary Marine Regions EEZ shapefile to inspect their columns and output the first few rows.

```python
import pandas as pd
import geopandas as gpd
```

### 1. Primary Dataset: Global Fishing Watch (Welch et al. 2022)
Make sure you have downloaded the CSV file and placed it in the same directory as this notebook (or update the path).

```python
# Replace with your actual downloaded CSV file name/path
csv_file_path = 'ais_disabling_events.csv'

try:
    # Load the dataset
    df_ais = pd.read_csv(csv_file_path)
    
    # Print the columns
    print("Columns in GFW Dataset:")
    print(df_ais.columns.tolist())
    print("\n--- \n")
    
    # Display the first 5 rows
    display(df_ais.head())
except FileNotFoundError:
    print(f"File '{csv_file_path}' not found. Please update the path.")
```

### 2. Secondary Dataset: Marine Regions EEZ Shapefiles
Make sure you have extracted the downloaded shapefile folder and point the path to the `.shp` file.

```python
# Replace with the path to the extracted .shp file
shapefile_path = 'World_EEZ_v12_20231025.shp'

try:
    # Load the geospatial dataset
    gdf_eez = gpd.read_file(shapefile_path)
    
    # Print the columns
    print("Columns in EEZ Shapefile:")
    print(gdf_eez.columns.tolist())
    print("\n--- \n")
    
    # Display the first 5 rows
    display(gdf_eez.head())
except FileNotFoundError:
    print(f"File '{shapefile_path}' not found. Please update the path.")
```
